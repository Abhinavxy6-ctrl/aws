```python
# ============================================================
# SKYGUARD V2 - STEP 6
# INTERACTIVE INDIA ANOMALY MAP
# ============================================================

import pandas as pd
import numpy as np
import plotly.express as px
from IPython.display import display

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

BASE_DIR = r"C:\Users\abhin\sih"
CITY_FILE = BASE_DIR + r"\skyguard_city_anomaly_summary.csv"

print("=" * 60)
print("       SKYGUARD V2 - INTERACTIVE INDIA ANOMALY MAP")
print("=" * 60)

# ------------------------------------------------------------
# Load city-level anomaly data
# ------------------------------------------------------------

city_df = pd.read_csv(CITY_FILE)

print(f"Cities loaded : {len(city_df):,}")
print(f"Columns       : {list(city_df.columns)}")

# ------------------------------------------------------------
# Load coordinates from original dataset
#
# We only need ONE coordinate per city.
# Therefore we read the original file in chunks and extract
# unique state/city/lat/lon combinations.
# ------------------------------------------------------------

DATASET_FILE = r"C:\Users\abhin\Downloads\archive\Indian_Weather_Dataset.csv"

print("\nReading city coordinates from original dataset...")
print("(Only unique city coordinates are retained.)")

coordinates = []

for chunk in pd.read_csv(
    DATASET_FILE,
    usecols=["state", "city", "lat", "lon"],
    chunksize=2_000_000
):
    coords = chunk.dropna(
        subset=["state", "city", "lat", "lon"]
    )[["state", "city", "lat", "lon"]].drop_duplicates()

    coordinates.append(coords)

coordinates_df = pd.concat(
    coordinates,
    ignore_index=True
).drop_duplicates(
    subset=["state", "city"]
)

print(f"Unique city coordinates : {len(coordinates_df):,}")

# ------------------------------------------------------------
# Merge anomaly information + coordinates
# ------------------------------------------------------------

map_df = city_df.merge(
    coordinates_df,
    on=["state", "city"],
    how="left"
)

print("\n------------------------------------------------------------")
print("MAP DATA CHECK")
print("------------------------------------------------------------")

print(f"Total locations : {len(map_df):,}")
print(f"Missing latitude  : {map_df['lat'].isna().sum():,}")
print(f"Missing longitude : {map_df['lon'].isna().sum():,}")

# ------------------------------------------------------------
# Remove locations without coordinates
# ------------------------------------------------------------

map_df = map_df.dropna(
    subset=["lat", "lon"]
).copy()

# ------------------------------------------------------------
# Create anomaly-rate categories
# ------------------------------------------------------------

def classify_risk(rate):
    if rate >= 10:
        return "Critical"
    elif rate >= 5:
        return "High"
    elif rate >= 2:
        return "Warning"
    else:
        return "Normal"

map_df["risk_level"] = map_df[
    "anomaly_rate_percent"
].apply(classify_risk)

# ------------------------------------------------------------
# Hover information
# ------------------------------------------------------------

map_df["location"] = (
    map_df["city"] + ", " + map_df["state"]
)

# ------------------------------------------------------------
# Create interactive map
# ------------------------------------------------------------

fig = px.scatter_map(
    map_df,
    lat="lat",
    lon="lon",
    size="anomalies",
    color="anomaly_rate_percent",
    hover_name="location",
    hover_data={
        "lat": ":.4f",
        "lon": ":.4f",
        "anomaly_rate_percent": ":.2f",
        "anomalies": ":,",
        "avg_score": ":.3f",
        "risk_level": True
    },
    color_continuous_scale="Turbo",
    size_max=35,
    zoom=4.2,
    center={
        "lat": 22.5,
        "lon": 79.0
    },
    map_style="open-street-map",
    title="SkyGuard V2 — India Weather Station Anomaly Map"
)

fig.update_layout(
    height=750,
    margin=dict(l=0, r=0, t=60, b=0)
)

fig.show()

print("\n============================================================")
print("       MAP READY")
print("============================================================")
print(f"Locations displayed : {len(map_df):,}")
print("Map features:")
print(" • Location")
print(" • Anomaly rate")
print(" • Number of anomalies")
print(" • Average anomaly score")
print(" • Risk level")
print("============================================================")

print("\n✅ Step 6 completed successfully.")
```

```python
# ============================================================
# SKYGUARD V2 - STEP 7
# INTERACTIVE LOCATION FILTERS
# ============================================================

import pandas as pd
import numpy as np
import plotly.express as px
from IPython.display import display

print("=" * 60)
print("       SKYGUARD V2 - INTERACTIVE LOCATION ANALYSIS")
print("=" * 60)

# ------------------------------------------------------------
# CHECK MAP DATA
# ------------------------------------------------------------

if "map_df" not in globals():
    raise RuntimeError(
        "map_df not found. Please run Step 6 first."
    )

analysis_df = map_df.copy()

print(f"Total locations available : {len(analysis_df):,}")

# ------------------------------------------------------------
# RISK CLASSIFICATION
# ------------------------------------------------------------

def risk_level(rate):
    if rate >= 10:
        return "Critical"
    elif rate >= 5:
        return "High"
    elif rate >= 2:
        return "Warning"
    else:
        return "Normal"

analysis_df["risk_level"] = (
    analysis_df["anomaly_rate_percent"]
    .apply(risk_level)
)

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("\n------------------------------------------------------------")
print("RISK DISTRIBUTION")
print("------------------------------------------------------------")

risk_summary = (
    analysis_df["risk_level"]
    .value_counts()
    .reindex(
        ["Critical", "High", "Warning", "Normal"],
        fill_value=0
    )
)

for level, count in risk_summary.items():
    print(f"{level:10s}: {count:,}")

# ------------------------------------------------------------
# USER FILTERS
# ------------------------------------------------------------

print("\n------------------------------------------------------------")
print("AVAILABLE STATES")
print("------------------------------------------------------------")

states = sorted(
    analysis_df["state"].dropna().unique()
)

print(f"States available : {len(states)}")

# ------------------------------------------------------------
# SELECT STATE
# ------------------------------------------------------------

selected_state = input(
    "\nEnter state name or type ALL: "
).strip()

if selected_state.upper() == "ALL":
    filtered_df = analysis_df.copy()
else:
    filtered_df = analysis_df[
        analysis_df["state"].str.lower()
        == selected_state.lower()
    ].copy()

# ------------------------------------------------------------
# SELECT MINIMUM ANOMALY RATE
# ------------------------------------------------------------

rate_input = input(
    "Minimum anomaly rate (%) [default 0]: "
).strip()

if rate_input == "":
    min_rate = 0.0
else:
    min_rate = float(rate_input)

filtered_df = filtered_df[
    filtered_df["anomaly_rate_percent"] >= min_rate
].copy()

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("\n============================================================")
print("FILTER RESULTS")
print("============================================================")

print(f"Locations matching filters : {len(filtered_df):,}")
print(f"Minimum anomaly rate       : {min_rate:.2f}%")

# ------------------------------------------------------------
# DISPLAY TOP LOCATIONS
# ------------------------------------------------------------

top_locations = (
    filtered_df
    .sort_values(
        "anomaly_rate_percent",
        ascending=False
    )
    .head(20)
)

print("\n------------------------------------------------------------")
print("TOP ANOMALOUS LOCATIONS")
print("------------------------------------------------------------")

display(
    top_locations[
        [
            "state",
            "city",
            "anomalies",
            "avg_score",
            "anomaly_rate_percent",
            "risk_level"
        ]
    ].reset_index(drop=True)
)

# ------------------------------------------------------------
# CREATE FILTERED MAP
# ------------------------------------------------------------

if len(filtered_df) > 0:

    filtered_df["location"] = (
        filtered_df["city"]
        + ", "
        + filtered_df["state"]
    )

    fig = px.scatter_map(
        filtered_df,
        lat="lat",
        lon="lon",
        size="anomalies",
        color="anomaly_rate_percent",
        hover_name="location",
        hover_data={
            "state": True,
            "city": True,
            "anomalies": ":,",
            "avg_score": ":.3f",
            "anomaly_rate_percent": ":.2f",
            "risk_level": True,
            "lat": ":.4f",
            "lon": ":.4f"
        },
        color_continuous_scale="Turbo",
        size_max=40,
        zoom=4.5,
        center={
            "lat": 22.5,
            "lon": 79.0
        },
        map_style="open-street-map",
        title="SkyGuard V2 — Filtered Anomaly Map"
    )

    fig.update_layout(
        height=700,
        margin=dict(
            l=0,
            r=0,
            t=60,
            b=0
        )
    )

    fig.show()

else:
    print("\n⚠️ No locations match the selected filters.")

print("\n============================================================")
print("       STEP 7 COMPLETE")
print("============================================================")
```

```python
import pandas as pd
from pathlib import Path

print("=" * 60)
print("       SKYGUARD V2 - CITY COORDINATE FILE")
print("=" * 60)

dataset_path = Path(r"C:\Users\abhin\Downloads\archive\Indian_Weather_Dataset.csv")
output_path = Path.cwd() / "skyguard_city_coordinates.csv"

print("\nReading city coordinates from original dataset...")
print("Dataset:", dataset_path)

# Only read the columns we need
coordinates = pd.read_csv(
    dataset_path,
    usecols=["state", "city", "lat", "lon"]
)

# Remove duplicate city/state combinations
coordinates = coordinates.drop_duplicates(
    subset=["state", "city"]
)

# Remove rows without coordinates
coordinates = coordinates.dropna(
    subset=["lat", "lon"]
)

# Save compact coordinate file
coordinates.to_csv(
    output_path,
    index=False
)

print("\n" + "-" * 60)
print("COORDINATE CHECK")
print("-" * 60)

print("Unique locations :", len(coordinates))
print("Missing latitude :", coordinates["lat"].isna().sum())
print("Missing longitude:", coordinates["lon"].isna().sum())

print("\nSaved:")
print(output_path)

print("\n" + "=" * 60)
print("✅ CITY COORDINATE FILE CREATED")
print("=" * 60)
```

```python
from pathlib import Path

app_path = Path.cwd() / "skyguard_dashboard.py"

app_code = r'''
import streamlit as st
import pandas as pd
import plotly.express as px
from pathlib import Path

# ============================================================
# SKYGUARD V2 - AI WEATHER STATION ANOMALY DASHBOARD
# ============================================================

st.set_page_config(
    page_title="SkyGuard V2",
    page_icon="🌦️",
    layout="wide"
)

BASE_DIR = Path(__file__).parent


# ============================================================
# LOAD DATA
# ============================================================

@st.cache_data
def load_data():

    city = pd.read_csv(
        BASE_DIR / "skyguard_city_anomaly_summary.csv"
    )

    coordinates = pd.read_csv(
        BASE_DIR / "skyguard_city_coordinates.csv"
    )

    state = pd.read_csv(
        BASE_DIR / "skyguard_state_anomaly_summary.csv"
    )

    sensor = pd.read_csv(
        BASE_DIR / "skyguard_sensor_anomaly_summary.csv"
    )

    contribution = pd.read_csv(
        BASE_DIR / "skyguard_sensor_contributions.csv"
    )

    monthly = pd.read_csv(
        BASE_DIR / "skyguard_monthly_anomaly_summary.csv"
    )

    hourly = pd.read_csv(
        BASE_DIR / "skyguard_hourly_anomaly_summary.csv"
    )

    persistent = pd.read_csv(
        BASE_DIR / "skyguard_persistent_events.csv"
    )

    top = pd.read_csv(
        BASE_DIR / "skyguard_top_anomalies.csv"
    )

    # --------------------------------------------------------
    # MERGE CITY DATA WITH COORDINATES
    # --------------------------------------------------------

    city = city.merge(
        coordinates,
        on=["state", "city"],
        how="left"
    )

    return (
        city,
        state,
        sensor,
        contribution,
        monthly,
        hourly,
        persistent,
        top
    )


(
    city,
    state,
    sensor,
    contribution,
    monthly,
    hourly,
    persistent,
    top
) = load_data()


# ============================================================
# HEADER
# ============================================================

st.title("🌦️ SkyGuard V2")

st.subheader(
    "AI/ML-Based Intelligent Anomaly Detection "
    "for Automatic Weather Stations"
)

st.markdown(
    """
    **SkyGuard V2** uses an autoencoder-based anomaly detection
    system to identify unusual weather-station observations,
    sensor anomalies and persistent anomaly events.
    """
)

st.divider()


# ============================================================
# OVERALL METRICS
# ============================================================

TOTAL_RECORDS = 46_082_160
TOTAL_ANOMALIES = 1_056_291
ANOMALY_RATE = 2.29

CRITICAL = 62_622
HIGH = 246_475
WARNING = 747_194

col1, col2, col3, col4 = st.columns(4)

with col1:
    st.metric(
        "📊 Records Analysed",
        f"{TOTAL_RECORDS:,}"
    )

with col2:
    st.metric(
        "🚨 Anomalies Detected",
        f"{TOTAL_ANOMALIES:,}"
    )

with col3:
    st.metric(
        "⚠️ Anomaly Rate",
        f"{ANOMALY_RATE:.2f}%"
    )

with col4:
    st.metric(
        "🔴 Critical",
        f"{CRITICAL:,}"
    )


st.divider()


# ============================================================
# SIDEBAR
# ============================================================

st.sidebar.title("🔎 SkyGuard Filters")

states = [
    "All States"
] + sorted(
    state["state"].dropna().unique().tolist()
)

selected_state = st.sidebar.selectbox(
    "Select State",
    states
)


if selected_state == "All States":

    filtered_city = city.copy()

else:

    filtered_city = city[
        city["state"] == selected_state
    ].copy()


# ============================================================
# RISK DISTRIBUTION
# ============================================================

st.header("🚨 Risk Distribution")

risk_df = pd.DataFrame({
    "Severity": [
        "Normal",
        "Warning",
        "High",
        "Critical"
    ],
    "Records": [
        45_025_869,
        WARNING,
        HIGH,
        CRITICAL
    ]
})

fig_risk = px.bar(
    risk_df,
    x="Severity",
    y="Records",
    title="Overall Anomaly Severity"
)

st.plotly_chart(
    fig_risk,
    use_container_width=True
)


# ============================================================
# INDIA ANOMALY MAP
# ============================================================

st.header("🗺️ India Anomaly Map")

map_data = filtered_city.dropna(
    subset=["lat", "lon"]
).copy()

if len(map_data) > 0:

    fig_map = px.scatter_map(
        map_data,
        lat="lat",
        lon="lon",
        size="anomalies",
        color="anomaly_rate_percent",
        hover_name="city",
        hover_data={
            "state": True,
            "anomalies": True,
            "avg_score": ":.3f",
            "anomaly_rate_percent": ":.2f",
            "lat": False,
            "lon": False
        },
        zoom=4,
        height=650,
        title="SkyGuard V2 - Anomaly Locations",
        color_continuous_scale="Turbo"
    )

    fig_map.update_layout(
        map_style="open-street-map"
    )

    st.plotly_chart(
        fig_map,
        use_container_width=True
    )

else:

    st.warning(
        "No coordinate data available for the selected state."
    )


# ============================================================
# TOP LOCATIONS
# ============================================================

st.subheader(
    "🏆 Highest Anomaly-Rate Locations"
)

top_locations = (
    filtered_city
    .sort_values(
        "anomaly_rate_percent",
        ascending=False
    )
    .head(10)
)

st.dataframe(
    top_locations[
        [
            "state",
            "city",
            "total_records",
            "anomalies",
            "avg_score",
            "anomaly_rate_percent"
        ]
    ],
    use_container_width=True,
    hide_index=True
)


# ============================================================
# SENSOR ANALYSIS
# ============================================================

st.header("🌡️ Sensor Anomaly Analysis")

fig_sensor = px.bar(
    sensor,
    x="sensor",
    y="anomaly_count",
    title="Dominant Sensor in Anomalies"
)

fig_sensor.update_layout(
    xaxis_tickangle=-45
)

st.plotly_chart(
    fig_sensor,
    use_container_width=True
)


# ============================================================
# SENSOR CONTRIBUTION
# ============================================================

st.subheader(
    "🤖 Sensor Contribution to Reconstruction Error"
)

fig_contribution = px.bar(
    contribution,
    x="feature",
    y="contribution_percent",
    title="Sensor Contribution"
)

fig_contribution.update_layout(
    xaxis_tickangle=-45
)

st.plotly_chart(
    fig_contribution,
    use_container_width=True
)


# ============================================================
# MONTHLY ANALYSIS
# ============================================================

st.header("📅 Temporal Analysis")

fig_month = px.line(
    monthly,
    x="month",
    y="anomaly_rate_percent",
    markers=True,
    title="Monthly Anomaly Rate"
)

st.plotly_chart(
    fig_month,
    use_container_width=True
)


# ============================================================
# HOURLY ANALYSIS
# ============================================================

fig_hour = px.line(
    hourly,
    x="hour",
    y="anomaly_rate_percent",
    markers=True,
    title="Hourly Anomaly Rate"
)

st.plotly_chart(
    fig_hour,
    use_container_width=True
)


# ============================================================
# PERSISTENT ANOMALIES
# ============================================================

st.header("⏱️ Persistent Anomaly Events")

st.metric(
    "Persistent Anomaly Records",
    f"{len(persistent):,}"
)

st.dataframe(
    persistent.head(100),
    use_container_width=True,
    hide_index=True
)


# ============================================================
# TOP SEVERE ANOMALIES
# ============================================================

st.header("🚨 Top 20 Severe Anomalies")

st.dataframe(
    top,
    use_container_width=True,
    hide_index=True
)


# ============================================================
# FOOTER
# ============================================================

st.divider()

st.caption(
    "SkyGuard V2 • AI/ML Weather Station Anomaly Detection • "
    "Intel Arc XPU Accelerated"
)
'''

app_path.write_text(
    app_code,
    encoding="utf-8"
)

print("=" * 60)
print("       SKYGUARD V2 - DASHBOARD UPDATED")
print("=" * 60)
print()
print("✅ Dashboard code updated")
print()
print("File:")
print(app_path)
print()
print("Coordinate file:")
print(
    Path.cwd() / "skyguard_city_coordinates.csv"
)
print()
print("The dashboard now merges city anomaly data")
print("with latitude/longitude automatically.")
print("=" * 60)
```
